# 03 — Results Analysis

Load `outputs/ablations/summary.json` and plot metric vs. each ablation axis.


```python
import json, pandas as pd, matplotlib.pyplot as plt

rows = json.load(open('../outputs/ablations/summary.json'))
df = pd.DataFrame(rows)
df.head()
```

```python
# Identity scale vs cross-frame CLIP similarity
if 'identity_scales' in df and 'cross_frame_clip' in df:
    g = df.groupby('identity_scales')['cross_frame_clip'].mean()
    g.plot(marker='o'); plt.xlabel('identity_scale'); plt.ylabel('Cross-frame CLIP sim'); plt.grid(True)
```
