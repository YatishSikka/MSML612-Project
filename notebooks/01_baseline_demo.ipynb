# 01 — Baseline Demo

Run vanilla StoryDiffusion on a handful of PororoSV captions. Verifies env + GPU are working before training.


```python
import sys, torch
sys.path.insert(0, '..')
from models.baselines.storydiffusion import VanillaStoryDiffusion

pipe = VanillaStoryDiffusion(torch_dtype=torch.float16)
captions = [
    'Pororo is waving at the camera in a snowy forest',
    'Pororo is walking through the snow toward a cabin',
    'Pororo opens the door of the cabin and smiles',
    'Pororo sits by the fireplace inside the cabin',
    'Pororo falls asleep on the couch'
]
images = pipe.generate(captions)
for i, im in enumerate(images):
    im.save(f'baseline_{i}.png')
```
