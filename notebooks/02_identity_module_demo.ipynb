# 02 — Identity Module Demo

Generate stories with the identity cross-attention active vs disabled.


```python
import sys, torch
from PIL import Image
sys.path.insert(0, '..')
from models.story_pipeline import StoryPipeline, StoryInput

pipe = StoryPipeline(
    base_model='runwayml/stable-diffusion-v1-5',
    identity_encoder_name='ip_adapter_clip',
    identity_dim=1024, num_identity_tokens=4, identity_scale=0.5,
    torch_dtype=torch.float16,
)
ref = Image.open('pororo_ref.png')
captions = ['...', '...', '...', '...', '...']

for scale in [0.0, 0.5, 1.0]:
    pipe.set_identity_scale(scale)
    imgs = pipe.generate(StoryInput(captions=captions, reference_images=[ref]))
    for i, im in enumerate(imgs):
        im.save(f'scale{scale}_frame{i}.png')
```
